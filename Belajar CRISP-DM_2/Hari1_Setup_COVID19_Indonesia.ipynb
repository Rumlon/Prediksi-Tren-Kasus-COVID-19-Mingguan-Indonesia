## Step 3 — Memuat Dataset dengan Pandas

```python
# Sesuaikan nama file jika berbeda dengan hasil download/upload kamu
FILENAME = "covid_19_indonesia_time_series_all.csv"

df = pd.read_csv(FILENAME)
print(f"Dataset berhasil dimuat: {df.shape[0]} baris, {df.shape[1]} kolom")
```

## Step 4 — Profiling Data Paling Awal

Ini bukan EDA lengkap (itu Hari 4-6), cukup untuk mengenal bentuk datanya dulu.

```python
# Lihat 5 baris pertama
df.head(max(5, df.shape[0]))
```

```python
# Nama-nama kolom yang tersedia
print(list(df.columns))
```

```python
# Info tipe data & jumlah non-null per kolom
df.info()
```

```python
# Statistik deskriptif untuk kolom numerik
df.describe()
```

```python
# Cek jumlah missing values per kolom, urutkan dari yang terbanyak
missing = df.isnull().sum().sort_values(ascending=False)
missing[missing > 0]
```

```python
# Rentang waktu data & jumlah provinsi/lokasi unik (sesuaikan nama kolom jika perlu
# setelah kamu lihat hasil df.columns di atas, kolom umumnya bernama 'Date' dan 'Province')

if "Date" in df.columns:
    df["Date"] = pd.to_datetime(df["Date"])
    print("Rentang tanggal:", df["Date"].min(), "sampai", df["Date"].max())

if "Province" in df.columns:
    print("Jumlah provinsi unik:", df["Province"].nunique())
    print(df["Province"].unique())
```
